```python
import pickle
from pathlib import Path

import numpy as np
import plotly.graph_objects as go
from plotly.colors import qualitative

# Predefined dash styles to cycle through
DASH_STYLES = ["solid", "dash", "dot", "dashdot", "longdash", "longdashdot"]

def plot_pr_curve_plotly(px, py, ap, names, prefix=None):
    py = np.stack(py, axis=1)
    colors = qualitative.Plotly  # e.g., 10-color colorblind-friendly palette

    fig = go.Figure()

    for i, y in enumerate(py.T):
        color = colors[i % len(colors)]  
        dash_style = DASH_STYLES[i % len(DASH_STYLES)]
        best_idx = np.argmax(y)
        best_val = y[best_idx]
        best_px = px[best_idx]
        label = f"{names[i]} AP={ap[i, 0]:.3f} (best={best_val:.2f} at {best_px:.2f})"

        fig.add_trace(go.Scatter(
            x=px, y=y, mode="lines", name=label,
            line=dict(color=color, dash=dash_style)
        ))
        fig.add_trace(go.Scatter(
            x=[best_px], y=[best_val],
            mode="markers",
            marker=dict(color=color, size=5),
            showlegend=False
        ))

    # Mean curve (always solid black)
    mean_curve = py.mean(1)
    best_idx = np.argmax(mean_curve)
    best_val = mean_curve[best_idx]
    best_px = px[best_idx]

    fig.add_trace(go.Scatter(
        x=px, y=mean_curve, mode="lines",
        name=f"all classes mAP@0.5={ap[:, 0].mean():.3f} (best={best_val:.2f} at {best_px:.2f})",
        line=dict(width=4, color='black', dash="solid")
    ))
    fig.add_trace(go.Scatter(
        x=[best_px], y=[best_val],
        mode="markers", marker=dict(color='black', size=5),
        showlegend=False
    ))

    fig.update_layout(title=f"{prefix} Precision-Recall Curve",
                      xaxis_title="Recall", yaxis_title="Precision")
    fig.show()

    fig.write_html(f"{prefix}_PR_curve.html")


def plot_metric_curve_plotly(px, py, names, ylabel="Metric", prefix=None):
    colors = qualitative.Plotly  
    fig = go.Figure()

    for i, y in enumerate(py):
        color = colors[i % len(colors)]
        dash_style = DASH_STYLES[i % len(DASH_STYLES)]
        best_idx = np.argmax(y)
        best_val = y[best_idx]
        best_px = px[best_idx]
        label = f"{names[i]} (best={best_val:.2f} at {best_px:.2f})"

        fig.add_trace(go.Scatter(
            x=px, y=y, mode="lines", name=label,
            line=dict(color=color, dash=dash_style)
        ))
        fig.add_trace(go.Scatter(
            x=[best_px], y=[best_val],
            mode="markers",
            marker=dict(color=color, size=5),
            showlegend=False
        ))

    mean_curve = np.mean(py, axis=0)
    best_idx = np.argmax(mean_curve)
    best_val = mean_curve[best_idx]
    best_px = px[best_idx]

    fig.add_trace(go.Scatter(
        x=px, y=mean_curve, mode="lines",
        name=f"all classes (best={best_val:.2f} at {best_px:.2f})",
        line=dict(width=3, color='black', dash="solid")
    ))
    fig.add_trace(go.Scatter(
        x=[best_px], y=[best_val],
        mode="markers", marker=dict(color='black', size=5),
        showlegend=False
    ))

    fig.update_layout(
        title=f"{prefix} {ylabel} vs Confidence",
        xaxis_title="Confidence",
        yaxis_title=ylabel
    )
    fig.show()

    fig.write_html(f"{prefix}_{ylabel}_curve.html")



# ✅ Load metrics.pkl
pkl_path = Path("/Users/victoriaoberascher/Desktop/class-thres/IR/metrics.pkl")  # update this path as needed
with open(pkl_path, "rb") as f:
    d = pickle.load(f)

px, py, ap, names = d["px"], d["py"], d["ap"], d["names"]
p, r, f1 = d["precision"], d["recall"], d["f1"]

# ✅ Plot each
plot_pr_curve_plotly(px, py, ap, names, prefix="IR")
plot_metric_curve_plotly(px, f1, names, ylabel="F1", prefix="IR")
plot_metric_curve_plotly(px, p, names, ylabel="Precision", prefix="IR")
plot_metric_curve_plotly(px, r, names, ylabel="Recall", prefix="IR")
```
